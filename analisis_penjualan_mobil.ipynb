```python
data = data.drop(['S.No.'], axis=1)
data.info()
```

```python
from datetime import date
date.today().year
data['Car_Age'] = date.today().year - data['Year']
data.head()
```

```python
data['Brand'] = data.Name.str.split().str.get(0)
data['Model'] = data.Name.str.split().str.get(1) + data.Name.str.split().str.get(2)
data[['Name', 'Brand', 'Model']]
```

```python
print(data.Brand.unique())
print(data.Brand.nunique())
```

```python
searchfor = ['Isuzu', 'ISUZU', 'Mini', 'Land']
data[data.Brand.str.contains('|'.join(searchfor))].head(5)
```

```python
data["Brand"].replace({"ISUZU": "Isuzu", "Mini": "Mini Cooper", "Land": "Land Rover"}, inplace=True)
```

```python
import numpy as np

dataobject = data.select_dtypes(include=['object']).columns
datanumeric = data.select_dtypes(include=np.number).columns.tolist()

print("Categorical Variables:")
print(dataobject)
print("Numerical Variables:")
print(datanumeric)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

for col in datanumeric:
    print(col)
    print('Skew :', round(data[col].skew(), 2))
    plt.figure(figsize=(15, 4))
    
    plt.subplot(1, 2, 1)
    data[col].hist(grid=False)
    plt.ylabel('count')
    
    plt.subplot(1, 2, 2)
    sns.boxplot(x=data[col])
    
    plt.show()
```

```python
fig, axes = plt.subplots(3, 2, figsize = (20, 20))
fig.suptitle('Bar plot for all categorical variables in the dataset')

sns.countplot(ax = axes[0, 0], x = 'Fuel_Type', data = data, color = 'coral',
              order = data['Fuel_Type'].value_counts().index)
sns.countplot(ax = axes[0, 1], x = 'Transmission', data = data, color = 'violet',
              order = data['Transmission'].value_counts().index)
sns.countplot(ax = axes[1, 0], x = 'Owner_Type', data = data, color = 'Aqua',
              order = data['Owner_Type'].value_counts().index)
sns.countplot(ax = axes[1, 1], x = 'Location', data = data, color = 'Crimson',
              order = data['Location'].value_counts().index)
sns.countplot(ax = axes[2, 0], x = 'Brand', data = data, color = 'Sienna',
              order = data['Brand'].head(20).value_counts().index)
sns.countplot(ax = axes[2, 1], x = 'Model', data = data, color = 'lime',
              order = data['Model'].head(15).value_counts().index)

axes[1][1].tick_params(labelrotation=45)
axes[2][0].tick_params(labelrotation=70)
axes[2][1].tick_params(labelrotation=90)
```
